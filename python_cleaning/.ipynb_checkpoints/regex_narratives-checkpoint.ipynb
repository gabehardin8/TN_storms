```python
details_cleaned['CZ_NAME'] = details_cleaned['CZ_NAME'].str.replace('/SMOKY MOUNTAINS','').str.replace('NORTHWEST ','').str.replace('SOUTHEAST ', '').str.replace('EAST ','').str.replace('NORTH ', '').str.replace('WEST', '').str.replace('SOUTH ','').str.replace(' POLK','POLK')
```

```python
details_cleaned
```

```python
text = details_cleaned['EVENT_NARRATIVE'].astype(str).str.lower()

from collections import Counter

stopwords = {'the','a','an','of','and','or','to','in','on','at','was','were',
             'is','are','be','been','with','by','from','as','it','this','that'}

counter = Counter()
for t in text:
    words = re.findall(r"[a-z']+",t)
    for w in words:
        if w not in stopwords:
             counter[w] += 1

word_count = pd.DataFrame(counter.most_common(325), columns=['word','count'])
word_count
```

```python
percentile_df = severity_df.copy() # if you want old versions of a dataframe, be sure to use .copy(). Otherwise both versions are updated with subsequent code.
percentile_df.head()
```

```python
severity_df['DEATHS_DIRECT']
```

```python
severity_df
```

```python
severity_metrics = ['INJURIES_DIRECT', 'INJURIES_INDIRECT', 'DEATHS_DIRECT',
       'DEATHS_INDIRECT', 'DAMAGE_PROPERTY', 'DAMAGE_CROPS',
       'road_closures', 'power_outage']


for metric in severity_metrics:
    percentile_df[f'{metric}'] = severity_df[f'{metric}'].astype(float)
    percentile_df[f'{metric}'] = percentile_df[f'{metric}'].rank(pct=True)
    

percentile_df[['INJURIES_DIRECT', 'INJURIES_INDIRECT', 'DEATHS_DIRECT',
       'DEATHS_INDIRECT', 'DAMAGE_PROPERTY', 'DAMAGE_CROPS',
       'road_closures', 'power_outage']]
```

```python
severity_df
```

```python
severity_df['DAMAGE_PROPERTY']
```

```python
percentile_df['total_severity'] = percentile_df['INJURIES_DIRECT'] + percentile_df['INJURIES_INDIRECT'] + percentile_df['DEATHS_DIRECT'] +percentile_df['DEATHS_INDIRECT'] +percentile_df['DAMAGE_PROPERTY'] +percentile_df['DAMAGE_CROPS'] +percentile_df['road_closures'] +percentile_df['power_outage']
```

```python
percentile_df['total_severity'] = percentile_df['total_severity'].rank(pct=True)
percentile_df['total_severity']
```

```python
percentile_df[['EVENT_ID','total_severity']].head()
```

```python
bi_import = pd.merge(severity_df,percentile_df[['EVENT_ID','total_severity']], how = 'inner', on = 'EVENT_ID')
bi_import.head()
```

```python
bi_import['CZ_NAME'] = bi_import['CZ_NAME'] +", TENNESSEE"
```
