```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import sqlalchemy as sql
import seaborn as sns
import os
```

```python
#saving csvs to variables

fatalities_2016 = pd.read_csv('StormEvents_fatalities-ftp_v1.0_d2016_c20260323.csv')
fatalities_2017 = pd.read_csv('StormEvents_fatalities-ftp_v1.0_d2017_c20260519.csv')
fatalities_2018 = pd.read_csv('StormEvents_fatalities-ftp_v1.0_d2018_c20260323.csv')
fatalities_2019 = pd.read_csv('StormEvents_fatalities-ftp_v1.0_d2019_c20260323.csv')
fatalities_2020 = pd.read_csv('StormEvents_fatalities-ftp_v1.0_d2020_c20260323.csv')
fatalities_2021 = pd.read_csv('StormEvents_fatalities-ftp_v1.0_d2021_c20260323.csv')
fatalities_2022 = pd.read_csv('StormEvents_fatalities-ftp_v1.0_d2022_c20260625.csv')
fatalities_2023 = pd.read_csv('StormEvents_fatalities-ftp_v1.0_d2023_c20260323.csv')
fatalities_2024 = pd.read_csv('StormEvents_fatalities-ftp_v1.0_d2024_c20260421.csv')
fatalities_2025 = pd.read_csv('StormEvents_fatalities-ftp_v1.0_d2025_c20260526.csv')
```

```python
#combining separate csv records of storm fatalities

fatalities_decade = pd.concat(
    [fatalities_2016, fatalities_2017, fatalities_2018, fatalities_2019, fatalities_2020, fatalities_2021, fatalities_2022, fatalities_2023, fatalities_2024, fatalities_2025]
)
```

```python
fatalities_decade
```

```python
#ensuring there are no duplicate records the the combined dataframe
fatalities_decade.drop_duplicates().shape
```

```python
fatalities_decade = fatalities_decade.reset_index(drop=True)
fatalities_decade.head()
```

```python
#setting storm detail csvs to variables

details_2016 = pd.read_csv('StormEvents_details-ftp_v1.0_d2016_c20260323.csv')
details_2017 = pd.read_csv('StormEvents_details-ftp_v1.0_d2017_c20260519.csv')
details_2018 = pd.read_csv('StormEvents_details-ftp_v1.0_d2018_c20260323.csv')
details_2019 = pd.read_csv('StormEvents_details-ftp_v1.0_d2019_c20260323.csv')
details_2020 = pd.read_csv('StormEvents_details-ftp_v1.0_d2020_c20260323.csv')
details_2021 = pd.read_csv('StormEvents_details-ftp_v1.0_d2021_c20260323.csv')
details_2022 = pd.read_csv('StormEvents_details-ftp_v1.0_d2022_c20260625.csv')
details_2023 = pd.read_csv('StormEvents_details-ftp_v1.0_d2023_c20260323.csv')
details_2024 = pd.read_csv('StormEvents_details-ftp_v1.0_d2024_c20260421.csv')
details_2025 = pd.read_csv('StormEvents_details-ftp_v1.0_d2025_c20260526.csv')
```

```python
#combining separate csv records of storm details

details_decade = pd.concat(
    [details_2016 ,details_2017 ,details_2018 ,details_2019 ,details_2020 ,details_2021 ,details_2022 ,details_2023 ,details_2024 ,details_2025]
)
```

```python
details_decade
```

```python
details_decade.drop_duplicates().shape
```

```python
# filtering storm details to only include records from Tennessee

tn_details_decade = details_decade.loc[details_decade['STATE'] == 'TENNESSEE']
tn_details_decade.shape
```

```python
tn_details_decade = tn_details_decade.reset_index(drop=True)
```

```python
# inner joining fatality records with storm details to filter only fatalities that occurred in TN

tn_fatalities_decade = pd.merge(tn_details_decade, fatalities_decade, how = 'inner', on = 'EVENT_ID')
tn_fatalities_decade.head()
```

```python
tn_fatalities_decade.columns
```

```python
fatalities_decade.columns
```

```python
#subsetting to only include columns from the origian fatality records

tn_fatalities_decade = tn_fatalities_decade[fatalities_decade.columns]
tn_fatalities_decade.head()
```

```python
#repeating the previous steps but for storm locations

locations_2016 = pd.read_csv('StormEvents_locations-ftp_v1.0_d2016_c20260707.csv')
locations_2017 = pd.read_csv('StormEvents_locations-ftp_v1.0_d2017_c20260707.csv')
locations_2018 = pd.read_csv('StormEvents_locations-ftp_v1.0_d2018_c20260707.csv')
locations_2019 = pd.read_csv('StormEvents_locations-ftp_v1.0_d2019_c20260707.csv')
locations_2020 = pd.read_csv('StormEvents_locations-ftp_v1.0_d2020_c20260707.csv')
locations_2021 = pd.read_csv('StormEvents_locations-ftp_v1.0_d2021_c20260707.csv')
locations_2022 = pd.read_csv('StormEvents_locations-ftp_v1.0_d2022_c20260707.csv')
locations_2023 = pd.read_csv('StormEvents_locations-ftp_v1.0_d2023_c20260707.csv')
locations_2024 = pd.read_csv('StormEvents_locations-ftp_v1.0_d2024_c20260707.csv')
locations_2025 = pd.read_csv('StormEvents_locations-ftp_v1.0_d2025_c20260707.csv')
```

```python
locations_decade = pd.concat(
    [locations_2016 ,locations_2017 ,locations_2018 ,locations_2019 ,locations_2020 ,locations_2021 ,locations_2022 ,locations_2023 ,locations_2024 ,locations_2025]
)
```

```python
locations_decade
```

```python
locations_decade.drop_duplicates().shape
```

```python
locations_decade = locations_decade.reset_index(drop = True)
```

```python
#again using the tn_details csvs to subset other dataframes to only include TN records
tn_locations_decade = pd.merge(tn_details_decade,locations_decade, how = 'inner', on = ['EVENT_ID','EPISODE_ID'])
tn_locations_decade.shape
```
