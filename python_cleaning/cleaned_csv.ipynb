```python
import pandas as pd
import numpy as np
import os
```

### This notebook cleans the csv containing storm details so that aggregations are easier

```python
# reading in the previously joined storm details csvs
tn_details_decade = pd.read_csv('tn_details_decade.csv')
tn_details_decade.head()
```

```python
tn_details_decade.columns
```

```python
#creating a new variable consisting of a subset of the "tn_details_decade" csv using only columns I thought may be useful
details_cleaned = tn_details_decade[['BEGIN_DATE_TIME','END_DATE_TIME','EPISODE_ID', 'EVENT_ID',
       'EVENT_TYPE', 'CZ_TYPE', 'CZ_FIPS', 'CZ_NAME',
       'WFO', 'CZ_TIMEZONE', 
       'INJURIES_DIRECT', 'INJURIES_INDIRECT', 'DEATHS_DIRECT',
       'DEATHS_INDIRECT', 'DAMAGE_PROPERTY', 'DAMAGE_CROPS', 'SOURCE',
       'MAGNITUDE', 'MAGNITUDE_TYPE', 'FLOOD_CAUSE', 'TOR_F_SCALE',
       'TOR_LENGTH', 'TOR_WIDTH', 'TOR_OTHER_WFO',
       'TOR_OTHER_CZ_FIPS', 'TOR_OTHER_CZ_NAME', 'BEGIN_RANGE',
       'BEGIN_AZIMUTH', 'BEGIN_LOCATION', 'END_RANGE', 'END_AZIMUTH',
       'END_LOCATION', 'BEGIN_LAT', 'BEGIN_LON', 'END_LAT', 'END_LON',
       'EPISODE_NARRATIVE', 'EVENT_NARRATIVE']]
details_cleaned.head()
```

```python
#looking to see if the datatypes are appropriately set for aggregations
details_cleaned.dtypes
```

```python
details_cleaned['DAMAGE_PROPERTY'] = (
    details_cleaned['DAMAGE_PROPERTY'].astype(str).str.replace('0.00K','0').str.replace('.','').str.replace('K','0').str.replace('M','0000').str.replace('B','0000000').str.replace('nan','0').astype(int)
)
details_cleaned['DAMAGE_PROPERTY'][10:20]

#cleaning a column containing prefixes for thousands, millions, and billions rather than a column of only numerical data like floats
```

```python
details_cleaned['DAMAGE_CROPS'] = (
    details_cleaned['DAMAGE_CROPS'].astype(str).str.replace('0.00K','0').str.replace('.','').str.replace('K','0').str.replace('M','0000').str.replace('B','0000000').str.replace('nan','0').astype(int)
)
details_cleaned['DAMAGE_CROPS'][25:50]
# same cleaning as above
```

```python
details_cleaned.head()
```

```python
details_cleaned['BEGIN_DATE'] = details_cleaned['BEGIN_DATE_TIME'].str[:9]
details_cleaned['BEGIN_TIME'] = details_cleaned['BEGIN_DATE_TIME'].str[10:]
details_cleaned['END_DATE'] = details_cleaned['END_DATE_TIME'].str[:9]
details_cleaned['END_TIME'] = details_cleaned['END_DATE_TIME'].str[10:]

details_cleaned.head()
#creating new columns that separates dates and times
```

```python
details_cleaned.columns
```

```python
details_cleaned = details_cleaned[['BEGIN_DATE', 'BEGIN_TIME', 'END_DATE', 'END_TIME','EPISODE_ID', 'EVENT_ID',
       'EVENT_TYPE', 'CZ_TYPE', 'CZ_FIPS', 'CZ_NAME', 'WFO', 'CZ_TIMEZONE',
       'INJURIES_DIRECT', 'INJURIES_INDIRECT', 'DEATHS_DIRECT',
       'DEATHS_INDIRECT', 'DAMAGE_PROPERTY', 'DAMAGE_CROPS', 'SOURCE',
       'MAGNITUDE', 'MAGNITUDE_TYPE', 'FLOOD_CAUSE', 'TOR_F_SCALE',
       'TOR_LENGTH', 'TOR_WIDTH', 'TOR_OTHER_WFO', 'TOR_OTHER_CZ_FIPS',
       'TOR_OTHER_CZ_NAME', 'BEGIN_RANGE', 'BEGIN_AZIMUTH', 'BEGIN_LOCATION',
       'END_RANGE', 'END_AZIMUTH', 'END_LOCATION', 'BEGIN_LAT', 'BEGIN_LON',
       'END_LAT', 'END_LON', 'EPISODE_NARRATIVE', 'EVENT_NARRATIVE']]

details_cleaned.head()

#subsetting the dataframe to hold only date columns that are separated out into data and times
```

```python
#the NOAA Storm Events database separates counties into separate regions that include cardinality prefixes. I stripped these off for the sake of getting countywide aggregations rather than regionwide aggregations
details_cleaned['CZ_NAME'] = (
    details_cleaned['CZ_NAME'].str.replace('/SMOKY MOUNTAINS','').str.replace('NORTHWEST ','').str.replace('SOUTHEAST ', '').str.replace('EAST ','').str.replace('NORTH ', '').str.replace('WEST ', '').str.replace('SOUTH ','')
)
```
